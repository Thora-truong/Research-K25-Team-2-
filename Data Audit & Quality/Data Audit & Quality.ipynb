```python
import os
import math
import numpy as np
import pandas as pd

# Đọc file Excel
sheets = pd.read_excel("Research_Project_Data.xlsx", sheet_name=None)

# Xem tên các sheet
print("Sheets:", list(sheets.keys()))

# Lấy 2 sheet cần kiểm tra
outbound = sheets["Outbound"]
branchmaster = sheets["BranchMaster"]

# Hàm kiểm tra dữ liệu
def check_data(df, sheet_name):

    print("\n" + "="*40)
    print(f"Sheet: {sheet_name}")
    print("="*40)

    # Xem thông tin dữ liệu
    print(df.info())

    # 1. Missing values
    print("\nMissing values:")
    print(df.isnull().sum())
    print("Total Missing:", df.isnull().sum().sum())

    # 2. Duplicate rows
    print("\nDuplicate rows:", df.duplicated().sum())

    # 3. Duplicate ID
    for col in ["DestinationRef", "ShipToID", "GroupID"]:
        if col in df.columns:
            print(f"\n{col}")
            print("Missing:", df[col].isnull().sum())
            print("Duplicate:", df[col].duplicated().sum())

    # 4. Invalid values
    if "Latitude" in df.columns:
        print("\nInvalid Latitude:",
              ((df["Latitude"] < -90) | (df["Latitude"] > 90)).sum())

    if "Longitude" in df.columns:
        print("Invalid Longitude:",
              ((df["Longitude"] < -180) | (df["Longitude"] > 180)).sum())

    # 5. Unit check (KG & CBM)
    if "BranchDeliveryVolumeInKG" in df.columns:
        print("Negative KG:",
              (df["BranchDeliveryVolumeInKG"] < 0).sum())

    if "BranchDeliveryVolumeInCBM" in df.columns:
        print("Negative CBM:",
              (df["BranchDeliveryVolumeInCBM"] < 0).sum())

# Chạy kiểm tra cho 2 sheet
check_data(outbound, "Outbound")
check_data(branchmaster, "BranchMaster")
```
